```python
def fcfs(requests, head):
    total = 0
    order = []

    for req in requests:
        total += abs(head - req)
        order.append(req)
        head = req

    return order, total


def sstf(requests, head):
    total = 0
    order = []
    requests = requests.copy()

    while requests:
        nearest = min(requests, key=lambda x: abs(head - x))
        total += abs(head - nearest)
        order.append(nearest)
        head = nearest
        requests.remove(nearest)

    return order, total


requests = [0, 30, 41, 62, 14]
head = 20

fcfs_order, fcfs_seek = fcfs(requests, head)
sstf_order, sstf_seek = sstf(requests, head)

print("FCFS Disk Scheduling")
print("Order:", fcfs_order)
print("Total Seek Time:", fcfs_seek)

print()

print("SSTF Disk Scheduling")
print("Order:", sstf_order)
print("Total Seek Time:", sstf_seek)

print()

if sstf_seek < fcfs_seek:
    print("Better Output: SSTF is better than FCFS")
else:
    print("Better Output: FCFS is better than SSTF")
```
